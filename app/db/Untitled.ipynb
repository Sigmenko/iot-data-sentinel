```python
import numpy as np
import matplotlib.pyplot as plt

# Налаштовуємо полотно (наш 2D простір)
plt.figure(figsize=(10, 8))
plt.axhline(0, color='black', linewidth=2) # Вісь X
plt.axvline(0, color='black', linewidth=2) # Вісь Y
plt.grid(True, linestyle='--', alpha=0.7)

# ---------------------------------------------------
# 1. СТВОРЮЄМО ВЕКТОРИ (Як масиви чисел)
# ---------------------------------------------------
v1 = np.array([20.0, 40.0])  # Поточний стан: Темп 20°C, Вологість 40%
v2 = np.array([10.0, -15.0]) # Зміна погоди: Тепліше на 10°C, сухіше на 15%

# Математика додавання (Векторизація)
v3_add = v1 + v2

# Математика скалярного множення (Збільшуємо вектор v1 в 1.5 раза)
# ТВОЄ ЗАВДАННЯ: спробуй змінити 1.5 на -1
v4_scale = v1 * -1

# ---------------------------------------------------
# 2. ФУНКЦІЯ ДЛЯ МАЛЮВАННЯ СТРІЛОЧОК (Як у фізиці)
# ---------------------------------------------------
def plot_vector(vec, origin=[0,0], color='b', label=''):
    """Малює вектор від заданої точки origin"""
    plt.quiver(*origin, *vec, color=color, angles='xy', scale_units='xy', scale=1, label=label)

# ---------------------------------------------------
# 3. МАЛЮЄМО ДОДАВАННЯ ВЕКТОРІВ (Правило трикутника)
# ---------------------------------------------------
plot_vector(v1, color='blue', label='v1 (Поточний стан: 20°C, 40%)')
plot_vector(v2, origin=v1, color='green', label='v2 (Зміна погоди)')
plot_vector(v3_add, color='red', label='v3 = v1 + v2 (Новий стан)')

# ---------------------------------------------------
# 4. МАЛЮЄМО СКАЛЯРНЕ МНОЖЕННЯ
# ---------------------------------------------------
# Малюємо його пунктиром, щоб побачити, що він лежить на тій самій прямій
plt.plot([0, v4_scale[0]], [0, v4_scale[1]], color='purple', linestyle=':', linewidth=4, 
         label=f'v4 = v1 * скаляр (Результат: {v4_scale})')

# Налаштування меж графіка та легенди
plt.xlim(-40, 60) # Розширив межі в мінус, щоб ти побачив розворот вектора!
plt.ylim(-60, 80)
plt.xlabel('Ось X (Температура °C)', fontsize=12)
plt.ylabel('Ось Y (Вологість %)', fontsize=12)
plt.title('Геометрична інтуїція: Додавання векторів та Скалярне множення', fontsize=14)
plt.legend(loc='upper left')
plt.show()
```
